```python
pipeline_options = PipelineOptions()

def process_data(data):
  global list_data
  list_data = data
with beam.Pipeline(options=pipeline_options) as pipeline:

    # Read CSV and parse lines into lists
    source_data = (
        pipeline
        | "Read CSV File" >> beam.io.ReadFromText('/content/Untitled document.txt') # skip_header_lines=1 -->> to exclude 1st rows
        | "Parse CSV to List" >> beam.Map(lambda line: line.split(','))
        #| "print the data " >> beam.Map(print)
    )

    list_data =source_data | "Collect Updated Data" >> beam.combiners.ToList()
    list_data | "print the data " >> beam.Map(process_data)



```
